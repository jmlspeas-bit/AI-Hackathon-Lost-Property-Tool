# Lost property at a city transit authority

Today you're working for the public transit authority of a mid-sized US city. The authority runs 38 bus routes and two light rail lines, and carries about 120,000 riders a day.

You'll be working with Denise, the lost property manager. Her team of three receives around 750 items a month, handed in by drivers, station staff, and riders. Each item is logged in a spreadsheet the next working morning, after the vehicle returns to its depot, and stored at the lost property office downtown.

## How lost property works today

Riders who lose something fill in a form on the authority's website. Denise's team works through the forms in order, searches the spreadsheet for a match, and replies within 72 hours. When they find a match, they contact the rider, check the item is theirs, and arrange collection or shipping.

Denise has described these problems with the current process:

- The office receives about 1,100 forms a month. Each takes a staff member about 12 minutes, which adds up to 220 hours a month.
- Riders wait up to 72 hours for a reply. That's too long for someone who has lost medication, house keys, or a phone they need for work.
- Riders describe items differently from how staff log them, so searches miss matches. Last year, only 28% of items went back to their owners.
- Some people claim items that aren't theirs, mostly phones and wallets.
- Forms that don't match anything sit in a separate list, and staff rarely have time to check them against items handed in later.

## Your task for today

Denise wants to replace the current process with a tool that gets riders a faster answer and frees up her team's time. Your team's task is to build an AI agent for her office that does this, using the materials below. What the agent does, who uses it, and how it fits into the office's work are for your team to work out.

## What does the board need to know about a new service?

Before the authority's board approves a new tool, Denise needs answers to these concerns:

- **Cost.** The board will want to know what the tool costs to run each month, compared with the staff time it replaces.
- **Data protection.** Returning items means collecting riders' names, contact details, and addresses. Many items also carry personal information, such as IDs, bank cards, and medication labels.
- **Fraud.** A faster process mustn't make it easier to claim someone else's property.
- **Mistakes.** The board will ask what happens when the tool gets something wrong, and who is responsible.
- **Staff capacity.** The office can't hire more staff this year.

- **No match yet:** The report is retained for future matching when new items are logged.

````python
# Loads the found items, lost reports, claims, and route files into pandas tables

import pandas as pd
import re
from pathlib import Path
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Load the challenge files
found_items = pd.read_csv("found_items.csv")
lost_reports = pd.read_csv("lost_reports.csv")
claims = pd.read_csv("claims.csv")
routes = pd.read_csv("routes_and_vehicles.csv")

print("Found items:", found_items.shape)
print("Lost reports:", lost_reports.shape)
print("Claims:", claims.shape)
print("Routes:", routes.shape)
````

````python
# Identifies which found item columns contain useful information for matching rider descriptions

search_columns = [
    "category",
    "description",
    "color",
    "brand",
    "found_on",
    "notes"
]
````

````python
#Combines each item's category, description, color, brand, route and notes into one searchable text field

found_items["search_text"] = (
    found_items[search_columns]
    .fillna("")
    .astype(str)
    .agg(" ".join, axis=1)
    .str.lower()
)

print(found_items[["item_id", "search_text"]].head())
````

````python
# Converts the searchable item text into numerical patterns that python can compare

vectorizer = TfidfVectorizer(
    stop_words="english",
    ngram_range=(1, 2)
)

item_vectors = vectorizer.fit_transform(
    found_items["search_text"]
)

print("Search index created.")
print("Number of indexed items:", item_vectors.shape[0])
````

````python
# Create a reusuable function that ranks found items by general text similiarity to a riders message

def find_matches(rider_message, top_k=5):
    """
    Search found items and return the strongest text matches.
    """
    
    # Convert the rider's message into the same numerical format
    # used for the found-item records.
    query_vector = vectorizer.transform([rider_message.lower()])
    
    # Compare the rider's message against every found item.
    similarity_scores = cosine_similarity(
        query_vector,
        item_vectors
    ).flatten()
    
    # Make a copy so we do not change the original data.
    results = found_items.copy()
    
    # Add the calculated score to each found item.
    results["match_score"] = similarity_scores
    
    # Put the strongest matches first.
    results = results.sort_values(
        by="match_score",
        ascending=False
    )
    
    # Show only the fields useful for reviewing a possible match.
    display_columns = [
        "item_id",
        "category",
        "description",
        "found_on",
        "date_found",
        "status",
        "notes",
        "match_score"
    ]
    
    return results[display_columns].head(top_k)
````

````python
#Stores the rider's pocket-knife request as the test input

m10_message = """
I left my pocket knife on the 5 bus on Saturday.
Red Swiss Army one, it was my dad's. Can I pick it up?
"""

m10_matches = find_matches(m10_message, top_k=5)

m10_matches
````

````python
#Improved search function and stores the results.

def find_matches_improved(rider_message, top_k=5):
    """
    Search found items using text similarity plus
    important lost-property keywords.
    """
    
    query = rider_message.lower()
    
    # Start with the original TF-IDF similarity score.
    query_vector = vectorizer.transform([query])
    scores = cosine_similarity(
        query_vector,
        item_vectors
    ).flatten()
    
    results = found_items.copy()
    results["match_score"] = scores
    
    # Combine the rider's message with important searchable fields.
    results["full_text"] = (
        results["category"].fillna("").astype(str) + " " +
        results["description"].fillna("").astype(str) + " " +
        results["color"].fillna("").astype(str) + " " +
        results["brand"].fillna("").astype(str) + " " +
        results["found_on"].fillna("").astype(str)
    ).str.lower()
    
    # Important item concepts and their alternate descriptions.
    concept_groups = [
        ["knife", "swiss army", "victorinox", "multi-tool", "weapon"],
        ["phone", "iphone", "mobile"],
        ["laptop", "macbook", "computer"],
        ["insulin", "medication", "medical"],
        ["wallet", "purse"],
        ["backpack", "bag"],
        ["wedding dress", "garment bag"],
        ["watch"],
        ["airpods", "earbuds", "headphones"]
    ]
    
    # Boost results when the rider and found item refer
    # to the same important item concept.
    for group in concept_groups:
        rider_mentions_concept = any(term in query for term in group)
        
        if rider_mentions_concept:
            results["concept_match"] = results["full_text"].apply(
                lambda text: any(term in text for term in group)
            )
            
            results.loc[
                results["concept_match"],
                "match_score"
            ] += 0.50
    
    # Boost results when the route or line matches.
    route_match = re.search(
        r"(route\s*\d+|\b\d+\s*bus|red line|blue line)",
        query
    )
    
    if route_match:
        rider_route = route_match.group(1).replace(" ", "")
        
        results["route_match"] = results["found_on"].apply(
            lambda value: rider_route in str(value).lower().replace(" ", "")
        )
        
        results.loc[
            results["route_match"],
            "match_score"
        ] += 0.30
    
    return results.sort_values(
        by="match_score",
        ascending=False
    )[[
        "item_id",
        "category",
        "description",
        "found_on",
        "date_found",
        "status",
        "notes",
        "match_score"
    ]].head(top_k)
````

````python
# Runs the basic matching function and returns the five highest-ranked results

m10_matches_improved = find_matches_improved(
    m10_message,
    top_k=5
)

m10_matches_improved
````

````python
# Creates a stronger matching function that combines text similarity with item keywords and route matching

def apply_policy(match):
    """
    Determine how the lost-property office should handle
    a possible found-item match.
    """
    
    category = str(match.get("category", "")).lower()
    description = str(match.get("description", "")).lower()
    status = str(match.get("status", "")).lower()
    notes = str(match.get("notes", "")).lower()
    value_band = str(match.get("value_band", "")).lower()
    
    full_text = f"{category} {description} {notes}"
    
    # Weapons and knives
    if (
        "weapon" in category
        or "knife" in full_text
        or "victorinox" in full_text
        or "swiss army" in full_text
        or "multi-tool" in full_text
    ):
        reference = None
        
        reference_match = re.search(
            r"ref\s+([A-Z]+-\d+-\d+)",
            notes,
            flags=re.IGNORECASE
        )
        
        if reference_match:
            reference = reference_match.group(1).upper()
        
        return {
            "decision": "Policy referral required",
            "reason": "This item is classified as a weapon.",
            "required_action": "Refer the rider to the Police Property Unit.",
            "staff_approval_required": True,
            "reference": reference
        }
    
    # Animals
    if (
        "animal" in category
        or "bearded dragon" in full_text
        or "pet" in full_text
    ):
        return {
            "decision": "Policy referral required",
            "reason": "Animals are handled by City Animal Services.",
            "required_action": "Refer the rider to City Animal Services.",
            "staff_approval_required": True,
            "reference": None
        }
    
    # Priority items
    priority_terms = [
        "medication",
        "insulin",
        "inhaler",
        "hearing aid",
        "walking cane",
        "walker",
        "wheelchair",
        "car seat"
    ]
    
    if any(term in full_text for term in priority_terms):
        return {
            "decision": "Priority handling",
            "reason": "This item may affect health, mobility, or child safety.",
            "required_action": "Confirm the match on the same working day.",
            "staff_approval_required": True,
            "reference": None
        }
    
    # High-value items
    high_value_terms = [
        "phone",
        "iphone",
        "laptop",
        "macbook",
        "tablet",
        "wallet",
        "purse",
        "cash",
        "watch",
        "jewelry",
        "hearing aid"
    ]
    
    if (
        value_band == "high"
        or any(term in full_text for term in high_value_terms)
    ):
        return {
            "decision": "Verification required",
            "reason": "This item requires stronger proof of ownership.",
            "required_action": (
                "Require an unlock, serial number, receipt, "
                "matching photo ID, or other policy-approved proof."
            ),
            "staff_approval_required": True,
            "reference": None
        }
    
    # Items already unavailable
    if status in ["disposed", "transferred"]:
        return {
            "decision": "Unavailable through normal collection",
            "reason": f"The item status is {status}.",
            "required_action": "Staff must provide the appropriate referral or explanation.",
            "staff_approval_required": True,
            "reference": None
        }
    
    # General case
    return {
        "decision": "Likely match",
        "reason": "The item description appears similar.",
        "required_action": "Verify the rider's ownership description.",
        "staff_approval_required": True,
        "reference": None
    }
````

````python
# Apply the policy to M10

best_m10_match = m10_matches_improved.iloc[0]

m10_policy = apply_policy(best_m10_match)

m10_policy
````

````python
# Applies the lost property rules to the folding knife match and generates the Police Property Unit referral

def process_rider_request(rider_message, top_k=5):
    """
    Process a rider message from search through policy review.
    """
    
    matches = find_matches_improved(
        rider_message,
        top_k=top_k
    )
    
    best_match = matches.iloc[0]
    policy_result = apply_policy(best_match)
    
    return {
        "rider_message": rider_message,
        "matches": matches,
        "best_match": best_match,
        "policy_result": policy_result
    }
````

````python
#Combines matching and policy review into one complete workflow for a rider's message

m10_result = process_rider_request(m10_message)

print("RIDER MESSAGE")
print(m10_result["rider_message"])

print("\nBEST MATCH")
print(m10_result["best_match"])

print("\nPOLICY RESULT")
for key, value in m10_result["policy_result"].items():
    print(f"{key}: {value}")
````

````python
# Generates a safe staff recommendation and rider response

def create_response(result):
    """
    Convert the matching and policy results into:
    1. An internal staff summary.
    2. A safe rider-facing response.
    """
    
    best_match = result["best_match"]
    policy = result["policy_result"]
    
    item_id = best_match["item_id"]
    status = best_match["status"]
    found_on = best_match["found_on"]
    score = best_match["match_score"]
    
    staff_summary = f"""
INTERNAL STAFF SUMMARY
----------------------
Likely item: {item_id}
Found on: {found_on}
Current status: {status}
Ranking score: {score:.3f}
Decision: {policy["decision"]}
Reason: {policy["reason"]}
Required action: {policy["required_action"]}
Staff approval required: {policy["staff_approval_required"]}
Reference: {policy["reference"]}
"""
    
    if policy["decision"] == "Policy referral required":
        rider_response = (
            "We found an item that may match your description. "
            "Because of the type of item, it is not handled through "
            "the transit authority's normal lost-property collection process. "
            f"Please contact the Police Property Unit and quote reference "
            f"{policy['reference']}."
        )
    
    elif policy["decision"] == "Priority handling":
        rider_response = (
            "We may have found an item matching your report. "
            "Because this may be a priority health or safety item, "
            "a member of our lost-property team will confirm the match "
            "as soon as possible today."
        )
    
    elif policy["decision"] == "Verification required":
        rider_response = (
            "We may have found an item matching your report. "
            "Because this is a high-value item, staff must verify ownership "
            "before arranging collection or shipping. Please provide an "
            "approved form of proof, such as an unlock, receipt, serial number, "
            "or matching photo ID."
        )
    
    else:
        rider_response = (
            "We may have found an item matching your report. "
            "Please provide additional identifying details so our team can "
            "verify ownership before arranging collection."
        )
    
    return staff_summary, rider_response
````

````python
# Generic high value rule was applied to cash and needs special case safety rules

def apply_special_handling(match, rider_message):
    """
    Check for special situations that require handling
    before applying the normal policy rules.
    """
    
    rider_text = rider_message.lower()
    item_text = (
        str(match.get("category", "")) + " " +
        str(match.get("description", "")) + " " +
        str(match.get("notes", ""))
    ).lower()
    
    # Cash requires exact amount and packaging verification.
    if "cash" in item_text or "money" in item_text:
        return {
            "decision": "Cash verification required",
            "reason": "Cash must be verified using the exact amount and packaging details.",
            "required_action": (
                "Ask the rider to confirm the exact amount, "
                "denominations, envelope, and any markings."
            ),
            "staff_approval_required": True,
            "reference": None
        }
    
    # A rider reporting a previous shipment needs a claim-status investigation.
    if (
        "mailed" in rider_text
        or "mail" in rider_text
        or "shipped" in rider_text
        or "shipping" in rider_text
    ):
        return {
            "decision": "Claim or shipment investigation required",
            "reason": "The rider reports that the item was previously approved for shipment.",
            "required_action": (
                "Search the claims record and verify the shipment or payment status "
                "before creating a new claim."
            ),
            "staff_approval_required": True,
            "reference": None
        }
    
    # Flag conflicting vehicle or item details for staff review.
    if (
        "toyota" in rider_text
        and "ford" in item_text
    ):
        return {
            "decision": "Possible conflicting match",
            "reason": "The rider reported a Toyota key, but the found item is a Ford key.",
            "required_action": (
                "Do not confirm the match; ask for additional details "
                "and review the item manually."
            ),
            "staff_approval_required": True,
            "reference": None
        }
    
    # Return None when no special handling rule applies.
    return None
````

````python
# Updated workflow to handle special case request

def process_rider_request_v2(rider_message, top_k=5):
    """
    Process a rider message using matching, special handling,
    and the standard lost-property policy.
    """
    
    matches = find_matches_improved(
        rider_message,
        top_k=top_k
    )
    
    best_match = matches.iloc[0]
    
    special_result = apply_special_handling(
        best_match,
        rider_message
    )
    
    if special_result is not None:
        policy_result = special_result
    else:
        policy_result = apply_policy(best_match)
    
    return {
        "rider_message": rider_message,
        "matches": matches,
        "best_match": best_match,
        "policy_result": policy_result
    }
````

````python
#Include request-level safety function

def handle_special_request(rider_message):
    """
    Identify requests that should be handled by privacy
    or records-management rules instead of item matching.
    """
    
    message = rider_message.lower()
    
    # Protect names and identifying information on found items.
    if (
        "list of all the wallets" in message
        or "list of all wallets and phones" in message
        or "names on them" in message
    ):
        return {
            "decision": "Privacy protection required",
            "reason": (
                "Names and identifying details recorded on found items "
                "are for authorized office use only."
            ),
            "required_action": (
                "Do not disclose names or item details. "
                "Verify the staff member's authorization and provide "
                "only an approved aggregate report."
            ),
            "staff_approval_required": True,
            "reference": None
        }
    
    # Handle rider requests to delete their personal information.
    if (
        "delete my details" in message
        or "delete my address" in message
        or "how long do you keep my details" in message
        or "rather you didn't keep my address" in message
    ):
        return {
            "decision": "Personal-data request",
            "reason": (
                "The rider is asking about retention or deletion "
                "of their personal information."
            ),
            "required_action": (
                "Delete the rider's details within 10 working days "
                "unless an open claim prevents deletion."
            ),
            "staff_approval_required": True,
            "reference": None
        }
    
    # Return None when the message is a normal lost-item request.
    return None
````

````python
#Creates a complete workflow

def process_rider_request_v3(rider_message, top_k=5):
    """
    Process a rider message through special-request checks,
    item matching, and policy review.
    """
    
    # Check privacy and records-management requests first.
    special_request = handle_special_request(rider_message)
    
    if special_request is not None:
        return {
            "rider_message": rider_message,
            "matches": pd.DataFrame(),
            "best_match": None,
            "policy_result": special_request
        }
    
    # Search for possible found-item matches.
    matches = find_matches_improved(
        rider_message,
        top_k=top_k
    )
    
    best_match = matches.iloc[0]
    
    # Check for cash, shipment, or conflicting details.
    special_handling = apply_special_handling(
        best_match,
        rider_message
    )
    
    if special_handling is not None:
        policy_result = special_handling
    else:
        policy_result = apply_policy(best_match)
    
    return {
        "rider_message": rider_message,
        "matches": matches,
        "best_match": best_match,
        "policy_result": policy_result
    }
````

````python
#Creation of universal response function

def create_response_v2(result):
    """
    Create an internal staff summary and a safe rider-facing response
    for both lost-item requests and special privacy requests.
    """
    
    policy = result["policy_result"]
    best_match = result["best_match"]
    
    # Handle privacy or personal-data requests that have no item match.
    if best_match is None:
        staff_summary = f"""
INTERNAL STAFF SUMMARY
----------------------
Request type: Special request
Decision: {policy["decision"]}
Reason: {policy["reason"]}
Required action: {policy["required_action"]}
Staff approval required: {policy["staff_approval_required"]}
"""
        
        if policy["decision"] == "Privacy protection required":
            rider_response = (
                "For privacy and security reasons, we cannot provide names "
                "or identifying details from lost-property records. "
                "Please submit an approved information request through "
                "the lost-property office."
            )
        
        elif policy["decision"] == "Personal-data request":
            rider_response = (
                "You may request deletion of your personal information. "
                "The authority processes deletion requests within 10 working days, "
                "unless the information is connected to an open claim."
            )
        
        else:
            rider_response = (
                "Your request has been sent to the lost-property office "
                "for staff review."
            )
        
        return staff_summary, rider_response
    
    # Handle normal lost-item requests with a possible match.
    item_id = best_match["item_id"]
    status = best_match["status"]
    found_on = best_match["found_on"]
    score = best_match["match_score"]
    
    staff_summary = f"""
INTERNAL STAFF SUMMARY
----------------------
Likely item: {item_id}
Found on: {found_on}
Current status: {status}
Ranking score: {score:.3f}
Decision: {policy["decision"]}
Reason: {policy["reason"]}
Required action: {policy["required_action"]}
Staff approval required: {policy["staff_approval_required"]}
Reference: {policy["reference"]}
"""
    
    if policy["decision"] == "Policy referral required":
        rider_response = (
            "We found an item that may match your description. "
            "Because of the type of item, it is not handled through "
            "the transit authority's normal lost-property process. "
            f"Please contact the appropriate authority and quote reference "
            f"{policy['reference']}."
        )
    
    elif policy["decision"] == "Priority handling":
        rider_response = (
            "We may have found an item matching your report. "
            "Because this may be a priority health or safety item, "
            "a member of our lost-property team will confirm the match "
            "as soon as possible today."
        )
    
    elif policy["decision"] == "Cash verification required":
        rider_response = (
            "We may have found an item matching your report. "
            "Before it can be released, staff will ask you to confirm "
            "the exact amount, denominations, envelope, and markings."
        )
    
    elif policy["decision"] == "Verification required":
        rider_response = (
            "We may have found an item matching your report. "
            "Because this is a high-value item, staff must verify ownership "
            "before arranging collection or shipping."
        )
    
    elif policy["decision"] == "Claim or shipment investigation required":
        rider_response = (
            "Your request has been sent to the lost-property office "
            "for staff review of the existing claim and shipment status."
        )
    
    elif policy["decision"] == "Possible conflicting match":
        rider_response = (
            "We found an item with some similar details, but additional "
            "information is needed before the match can be confirmed."
        )
    
    else:
        rider_response = (
            "We may have found an item matching your report. "
            "Please provide additional identifying details so staff "
            "can verify ownership."
        )
    
    return staff_summary, rider_response
````

````python
# Final demo function

def run_agent(rider_message, top_k=3):
    """
    Run the complete lost-property agent workflow.
    """
    
    result = process_rider_request_v3(
        rider_message,
        top_k=top_k
    )
    
    staff_summary, rider_response = create_response_v2(result)
    
    print("=" * 90)
    print("LOST-PROPERTY AGENT")
    print("=" * 90)
    
    print("\nRIDER MESSAGE")
    print("-" * 90)
    print(rider_message.strip())
    
    # Display candidate matches only when an item search was performed.
    if not result["matches"].empty:
        print("\nPOSSIBLE MATCHES FOR STAFF REVIEW")
        print("-" * 90)
        
        candidate_columns = [
            "item_id",
            "category",
            "description",
            "found_on",
            "date_found",
            "status",
            "match_score"
        ]
        
        display(result["matches"][candidate_columns])
    
    print("\nSTAFF RECOMMENDATION")
    print("-" * 90)
    print(staff_summary)
    
    print("\nRIDER-FACING RESPONSE")
    print("-" * 90)
    print(rider_response)
    
    return result
````

# Returning Lost Property Faster and More Safely

## The Problem

The lost-property team receives approximately 1,100 reports and 750 items each month, but staff spend 220 hours manually searching for matches. Riders wait up to 72 hours, and only 28% of items are returned to their owners.

`220 hours x $38/hour = $8,360/month in labor`
`12 minutes × ($38 ÷ 60) = $7.60 per report`

## Our Solution

We built an AI-assisted workflow that interprets rider descriptions, ranks likely found-item matches, applies lost-property policy, and gives staff a recommended next action. The goal is to enable the City Transit Authority to do the following:
- Faster searching
- More consistent matching
- Privacy protection
- Fraud prevention
- Human approval

### How the agent works
1. Rider message
2. Extract item details
3. Search found-item records
4. Rank possible matches
5. Apply policy and safety rules
6. Generate staff recommendation
7. Provide safe rider response

**The search component identifies possible matches, while a separate policy layer decides how the item may be handled.**
